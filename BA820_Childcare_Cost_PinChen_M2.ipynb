### Affordability Ratio


```python
import matplotlib.pyplot as plt
import seaborn as sns

# calculate the propotion of childcare costs to income
# define variable groups
age_groups = ['infant', 'toddler', 'preschool', 'sa']
care_types = ['mc', 'mfcc']
care_type_name = {'mc': 'center', 'mfcc': 'family'}

# create a subplot with 2 rows x 4 columns
fig, axes = plt.subplots(2, 4, figsize=(22, 10), sharex=True)
fig.suptitle('Affordability Ratios Across All Age Groups', fontsize=18)

for i, c_type in enumerate(care_types):
  for j, age in enumerate(age_groups):

    if age == 'sa':
      col_name = f'{c_type}{age}'
    else:
      col_name = f'{c_type}_{age}'

    ratio_col = f'{col_name}_ratio'  # for creating new columns

    # convert childcare cost to yearly price and divided by yearly income
    merged_df[ratio_col] = (merged_df[col_name] * 52) / merged_df['mhi_2018']

    # plot histogram
    sns.histplot(merged_df[ratio_col], kde=True, ax=axes[i, j], color='skyblue' if c_type=='mc' else 'orange')
    axes[i, j].set_title(f'{care_type_name[c_type]} - {age}')
    axes[i, j].set_xlabel('Affordability Ratio')
    axes[i, j].set_ylabel('Number of Counties')

    # mark the 10% affordability threshold
    axes[i, j].axvline(0.1, color='red', linestyle='--')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

These histograms visualize the affordability ratio of counties across four child age groups (infant, toddler, preschool, and school age) and two care types (Center-Based care and Family Childcare). The red dashed line represents the **10% affordability threshold** traditionally defined by the US Department of Health and Human Services (HHS, https://www.federalregister.gov/documents/2016/09/30/2016-22986/child-care-and-development-fund-ccdf-program). Across all eight categories, the area to the right of the red line represents a substantial proportion of counties where childcare is considered unaffordable.

Center-Based Care distributions are noticeably shifted further to the right compared to Family Childcare in child under 6 (infant, toddler, and pre-school) groups, indicating that center-based settings impose a substantially higher financial burden on median-income households.

As children transition from infant to school age status, the peak of the distribution moves slightly to the left, suggesting a decrease in cost pressure. However, the long tail persists, highlighting that for many counties, childcare remains a crisis regardless of the child's age.

### Affordability Deviation From Norms

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# define variable groups (the same as affordability ratio analysis)
age_groups = ['infant', 'toddler', 'preschool', 'sa']
care_types = ['mc', 'mfcc']
ratio_cols = ['mc_infant_ratio', 'mfcc_infant_ratio',
        'mc_toddler_ratio', 'mfcc_toddler_ratio',
        'mc_preschool_ratio', 'mfcc_preschool_ratio',
        'mcsa_ratio', 'mfccsa_ratio']

# calculate the average affordability ratio for each county across years
county_avg = merged_df.groupby(['county_name', 'state_name'])[ratio_cols].mean().reset_index()

# calculate the z-score of each county in its state
def zscore(x):
    return (x - x.mean()) / x.std()

z_cols = []
for col in ratio_cols:
    z_col = f'z_{col}'
    county_avg[z_col] = county_avg.groupby('state_name')[col].transform(zscore)
    z_cols.append(z_col)

# find the most anomalous counties by the max |z| across columns
county_avg['max_abs_z'] = county_avg[z_cols].abs().max(axis=1)  # pick the max |z| for each z_col

top_n = 20  # pick the top 20 counties
top_df = county_avg.nlargest(top_n, 'max_abs_z').copy()

# plot heatmap: rows = counties, columns = ratios
top_df['county_state'] = top_df['county_name'] + ', ' + top_df['state_name']
heatmap_data = top_df.set_index('county_state')[z_cols]

plt.figure(figsize=(14, 10))
sns.heatmap(heatmap_data, cmap='coolwarm', center=0, linewidths=0.3, linecolor='white')
plt.title(f'Deviation from State Norms (Z-scores) — Top {top_n} Most Anomalous Counties')
plt.xlabel('Care Type & Age Group (Z-score of Affordability Ratio within State)')
plt.ylabel('County, State')
plt.tight_layout()
plt.show()

# display the top 20 anomalies table
display_cols = ['county_name', 'state_name', 'max_abs_z'] + z_cols
top_anomalies = top_df[display_cols]
print(f"Top {top_n} Most Anomalous Counties (max absolute z-score across all ratios):")
print(top_anomalies.head(20))
```

The heatmap shows county-level deviations from state affordability norms using z-scores calculated from multi-year average affordability ratios, where positive values indicate higher childcare cost burdens relative to the state average.

Several large urban counties, such as Bronx County (NY), Philadelphia County (PA), Milwaukee County (WI), and Cook County (IL), exhibit consistently high positive z-scores across nearly all age groups and care types, suggesting a structurally elevated childcare affordability burden rather than age-specific or care-type-specific anomalies.

In contrast, some counties show more heterogeneous patterns, with deviations concentrated in particular age groups or care types, indicating localized mismatches rather than systemic stress.

Notably, Williamson County (TN) stands out with uniformly negative z-scores across all categories, reflecting a structurally lower childcare burden relative to its state norm and serving as a potential benchmark case.

### Correlation Analysis

```python
import seaborn as sns
import matplotlib.pyplot as plt

# create feature: gender gap in labor force participation
merged_df['lfp_gap'] = merged_df['flfpr_20to64'] - merged_df['mlfpr_20to64']  # has been created in M1 EDA

# define observed variances
analysis_vars = [
    'mc_infant_ratio', 'mfcc_infant_ratio',
    'mc_toddler_ratio', 'mfcc_toddler_ratio',
    'mc_preschool_ratio', 'mfcc_preschool_ratio',
    'flfpr_20to64', 'lfp_gap', 'pr_f'
]

# plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(merged_df[analysis_vars].corr(), annot=True, cmap='coolwarm', center=0, fmt=".2f")
plt.title('Correlation Matrix: Childcare Stress vs. Labor Outcomes')
plt.show()
```

This correlation analysis aims to identify relationships among columns and use the heatmap results to inform dimension selection in downstream unsupervised machine learning to avoid redundancy.

The heatmap indicates strong positive correlations among childcare affordability ratios across age groups and childcare types. Based on prior EDA results showing higher affordability ratios for infants Center-Based care, **mc_infant_ratio** was chosen as representative for childcare costs.

In addition, for the labor force participation dimension, I selected the gender gap in labor force participation (**lfp_gap**) and the female labor force participation rate (**flfpr_20to64**). These measures capture both gender disparities in labor participation at the county level and the relationship between childcare costs and women’s labor force participation.

I opted to use the poverty rate (**pr_f**) instead of median household income (mhi_2018) as a clustering feature to avoid the causality with affordability ratios.

### **(1) Can we use unsupervised clustering and anomaly detection to identify latent childcare crisis profiles in counties where costs, income, and labor participation exhibit extreme mismatch, and how do these clusters differ from regional economic norms?**

##### Data Pre-processing

```python
import numpy as np
import pandas as pd

# representative
rep_cols = [
    'mc_infant_ratio',
    'lfp_gap',
    'flfpr_20to64',
    'pr_f'
]

# prepare dataframe for clustering
id_cols = ['county_fips_code', 'county_name', 'state_name']
df_cluster = (merged_df[id_cols + rep_cols]
        .groupby(id_cols, as_index=False)[rep_cols]
        .mean())  # take the mean of each representative

print(f'df_cluster shape: {df_cluster.shape}')
df_cluster.info()
df_cluster.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# feature matrix
X = df_cluster[rep_cols].copy()

# build a pre-processing pipeline
prep_pipeline = Pipeline(steps=[('imputer', SimpleImputer(strategy='median')),  # fill missing values with median
              ('scaler', StandardScaler())])  # normalize: mean = 0, sd = 1

# transform data
Xs_std = prep_pipeline.fit_transform(X)

print(f'Xs_std shape: {Xs_std.shape}')
```

##### K-means Clustering

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

def kmeans_grid(Xs, k_list=range(2, 11), random_state=42):  # try k = 2 to 10
    rows = []
    labels_dict = {}

    for k in k_list:
        km = KMeans(n_clusters=k, random_state=random_state)
        labels = km.fit_predict(Xs)  # fit data and predict
        sse = km.inertia_  # SSE: sum of squared errors to cluster centers
        sil = silhouette_score(Xs, labels)  # silhouette score
        rows.append({'K': k, 'SSE': sse, 'silhouette_score': silhouette_score(Xs, labels)})
        labels_dict[k] = labels
    return pd.DataFrame(rows), labels_dict

# run on the scaler
res_std, labels_dict = kmeans_grid(Xs_std, range(2, 11))

display(res_std)
```

```python
# plot the Elbow and Silhouette figures

import matplotlib.pyplot as plt

# Elbow plot (SSE)
plt.figure(figsize=(6, 4))
plt.plot(res_std['K'], res_std['SSE'], marker='o')
plt.title('Elbow (SSE)')
plt.xlabel('K')
plt.ylabel('SSE (inertia)')
plt.show()

# Silhouette plot
plt.figure(figsize=(6, 4))
plt.plot(res_std['K'], res_std['silhouette_score'], marker='o')
plt.title('Silhouette')
plt.xlabel('K')
plt.ylabel('Silhouette score')
plt.show()
```

Using the K-means clustering method, I examined K values ranging from 2 to 10. The SSE smoothly declines along with increasing K. This transition suggests that there are no naturally rigid boundaries between clusters.

The highest silhouette score is at K=3, followed by K=4. When K≥5, the scores decrease significantly, indicating that additional clusters lead to poorer separation and more overlap. Although silhouette score is also high at K=2, it results in overly coarse clusters that obscure important heterogeneity in childcare affordability and female labor force outcomes.

Overall, these results support using K=2 as a simple baseline, while K=4 (with lower SSE than K=3, and a relatively high silhouette score) provides a more nuanced yet still well-separated typology for interpretation.

```python
# output cluster profile

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

rep_cols = ['mc_infant_ratio', 'lfp_gap', 'flfpr_20to64', 'pr_f']
k_to_plot = [2, 4]

for k in k_to_plot:
    labels = labels_dict[k]
    df_tmp = df_cluster.copy()  # temporary df for saving cluster labels
    df_tmp[f'cluster_k{k}'] = labels

    sizes = df_tmp[f'cluster_k{k}'].value_counts().sort_index()  # cluster size
    raw_profile = df_tmp.groupby(f'cluster_k{k}')[rep_cols].mean()  # take the mean for each feature

    # standardize: convert values to z-score (more comparable)
    df_std = df_tmp.copy()
    for c in rep_cols:
        df_std[c] = (df_std[c] - df_std[c].mean()) / df_std[c].std()

    # take the mean after standardization
    std_profile = df_std.groupby(f'cluster_k{k}')[rep_cols].mean()

    # set the mean as the threshold: > threshold (High); < threshold (Low)
    th_burden = raw_profile['mc_infant_ratio'].mean()
    th_flfp = raw_profile['flfpr_20to64'].mean()
    th_gap = raw_profile['lfp_gap'].mean()
    th_pov = raw_profile['pr_f'].mean()

    # name clusters
    cluster_names = {}
    for cl in raw_profile.index:
        burden_tag = 'High-burden' if raw_profile.loc[cl, 'mc_infant_ratio'] >= th_burden else 'Low-burden'
        flfp_tag = 'High-flfp' if raw_profile.loc[cl, 'flfpr_20to64'] >= th_flfp else 'Low-flfp'
        gap_tag = 'High-gap' if raw_profile.loc[cl, 'lfp_gap'] >= th_gap else 'Low-gap'
        pov_tag = 'High-poverty' if raw_profile.loc[cl, 'pr_f'] >= th_pov else 'Low-poverty'
        cluster_names[cl] = f'{burden_tag}, {flfp_tag}, {gap_tag}, {pov_tag})'

    # set rows' names and plot heatmap
    new_index = []
    for cl in std_profile.index:
        new_index.append(f'Cluster {cl} (n={sizes.loc[cl]})\n{cluster_names[cl]}')
    std_profile.index = new_index

    plt.figure(figsize=(10, 3 + 0.6 * len(std_profile)))
    sns.heatmap(std_profile, cmap='coolwarm', center=0, annot=True, fmt='.2f')
    plt.title(f'Cluster Profile Heatmap (standardized) — K={k}')
    plt.xlabel('Features (standardized)')
    plt.tight_layout()
    plt.show()

    # output raw means table
    raw_out = raw_profile.copy().round(2)
    raw_out.insert(0, 'cluster_size', sizes.values)
    raw_out.insert(1, 'cluster_name', [cluster_names[cl] for cl in raw_out.index])

    print(f'\n===== K = {k} : Cluster Summary (raw means) =====')
    display(raw_out)
```

For K=2, counties separate into two broad groups: one with higher infant childcare burden and poverty combined with lower female labor force participation, and another with lower burden and poverty and higher female labor force participation. This indicates a clear structural divide between high-stress and more advantaged childcare–labor market contexts.

For K=4, the clustering reveals more nuanced patterns. Compared to the corresponding average of the entire data, the results show that there is a large low-burden, high-participation, low-poverty group. In addition, there are counties with low childcare burden but low female labor participation and high poverty, suggesting that affordability alone does not guarantee labor market engagement. A distinct small cluster shows an extreme labor force participation gap, while another smaller cluster represents high childcare burden but relatively strong female labor participation, consistent with high-cost labor markets where childcare is expensive yet employment remains high.

##### Hierarchical Clustering

```python
from scipy.cluster.hierarchy import linkage, fcluster, dendrogram
import matplotlib.pyplot as plt
import pandas as pd

k_final = 4   # final choice of K

Z_ward = linkage(Xs_std, method='ward')  # build linkage matrix: try ward linkage
Z_avg = linkage(Xs_std, method='average')  # build linkage matrix: try average linkage

# plot dendrogram (Z_ward)
plt.figure(figsize=(14, 5))
dendrogram(Z_ward, truncate_mode='lastp', p=30, show_leaf_counts=True)  # only display the last part, and up to 30 clusters
cut_height = Z_ward[-(k_final-1), 2]  # mark the line with K clusters
plt.axhline(y=cut_height, c='red', linestyle='--', label=f'Cut for k={k_final}')
plt.legend()
plt.title(f'Hierarchical Dendrogram (ward linkage with cut at k = {k_final})')
plt.xlabel('Cluster (truncated)')
plt.ylabel('Distance')
plt.show()

# plot dendrogram (Z_avg)
plt.figure(figsize=(14, 5))
dendrogram(Z_avg, truncate_mode='lastp', p=30, show_leaf_counts=True)  # only display the last part, and up to 30 clusters
cut_height = Z_avg[-(k_final-1), 2]  # mark the line with K clusters
plt.axhline(y=cut_height, c='red', linestyle='--', label=f'Cut for k={k_final}')
plt.legend()
plt.title(f'Hierarchical Dendrogram (average linkage with cut at k = {k_final})')
plt.xlabel('Cluster (truncated)')
plt.ylabel('Distance')
plt.show()
```

The dendrograms visualize hierarchical clustering results using Ward and average linkage with a cut at K=4.

Under Ward linkage, several clear vertical jumps appear before the cut line, indicating relatively well-separated clusters with strong within-cluster similarity. This suggests the presence of distinct county profiles characterized by different levels of childcare affordability stress and female labor force participation.

In contrast, average linkage shows a poor separation between clusters. While a four-cluster solution is still feasible, the cluster boundaries are less distinct.

##### Compare K-means & Hierarchical Clustering

```python
from sklearn.cluster import KMeans

# create a dataframe to store clustering labels
z_df = df_cluster[id_cols + rep_cols].copy()

# cut the hierarchical tree into k_final clusters
z_df['cluster_hier_ward'] = fcluster(Z_ward, t=k_final, criterion='maxclust') - 1  # index starts from 0
z_df['cluster_hier_avg'] = fcluster(Z_avg, t=k_final, criterion='maxclust') - 1  # index starts from 0

# fit K-means to generate baseline labels
km = KMeans(n_clusters=k_final, random_state=42)
z_df['cluster_kmeans'] = km.fit_predict(Xs_std)

# compare consistency between K-means vs Hierarchical
print('Crosstab: K-means vs Hierarchical (Ward)')
display(pd.crosstab(z_df['cluster_kmeans'], z_df['cluster_hier_ward']))

print('Crosstab: K-means vs Hierarchical (Average)')
display(pd.crosstab(z_df['cluster_kmeans'], z_df['cluster_hier_avg']))
```

Based on the cross-table, the clustering results from K-means and Ward’s hierarchical method show partial alignment rather than a one-to-one correspondence. For example, K-means cluster 1 largely maps to Ward cluster 1 (1,325 observations), while K-means clusters 2 and 3 are distributed across multiple Ward clusters, indicating that the two methods define cluster boundaries differently for some counties.

In contrast, the Average linkage method assigns the majority of counties to a single cluster (cluster 1), resulting in an unclear clustering structure with limited discriminatory power.

### **(2) What are the underlying association patterns between the dominant childcare provider type, Center-Based Care vs. Family Childcare, and parental labor force participation, and can these patterns reveal where current infrastructure fails to support flexible work schedules?**

##### Data Pre-processing

```python
import numpy as np
import pandas as pd

# calculate county-level multi-year averages
id_cols = ['county_fips_code', 'county_name', 'state_name']

base_cols = [
    'mc_infant_ratio',
    'mfcc_infant_ratio',  # for calculating provider ratios and defining dominance
    'lfp_gap',
    'flfpr_20to64',
    'pr_f'
]

df_q2 = (merged_df[id_cols + ['study_year'] + base_cols]
      .groupby(id_cols, as_index=False)[base_cols]
      .mean()
)

# Screate provider dominance indicators
# provider_advantage > 0: Family Childcare burden higher than Center-Based care (Family Childcare less affordable)
# provider_advantage < 0: Center-Based care burden higher than Family Childcare (Center-Based care less affordable)
df_q2['provider_advantage'] = df_q2['mfcc_infant_ratio'] - df_q2['mc_infant_ratio']

print(f'df_q2 shape: {df_q2.shape}')
display(df_q2.head())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# define representative variables
feature_cols = [
    'mc_infant_ratio',
    'provider_advantage',  # relative provider dominance (Family Childcare vs Center-Based care)
    'lfp_gap',
    'flfpr_20to64',
    'pr_f'
]

X = df_q2[feature_cols].copy()

# build a pre-processing pipeline
prep = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
Xs = prep.fit_transform(X)
```

##### K-means Clustering

```python
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

k_list = range(2, 11)  # try K = 2 to 10
rows = []
labels_by_k = {}

for k in k_list:
    km = KMeans(n_clusters=k, random_state=42)
    labels = km.fit_predict(Xs)

    rows.append({
        'K': k,
        'SSE': km.inertia_,
        'silhouette': silhouette_score(Xs, labels)
    })
    labels_by_k[k] = labels

res_k = pd.DataFrame(rows)
display(res_k)
```

```python
import matplotlib.pyplot as plt

# Elbow plot (SSE)
plt.figure(figsize=(6, 4))
plt.plot(res_k['K'], res_k['SSE'], marker='o')
plt.title('Elbow Plot (SSE)')
plt.xlabel('K')
plt.ylabel('SSE (inertia)')
plt.show()

# Silhouette plot
plt.figure(figsize=(6, 4))
plt.plot(res_k['K'], res_k['silhouette'], marker='o')
plt.title('Silhouette')
plt.xlabel('K')
plt.ylabel('Silhouette Score')
plt.show()
```

In the Elbow (SSE) plot, SSE decreases most rapidly before K=4, then levels off, indicating that around 4 clusters are sufficient to capture the main structure.

The Silhouette plot also reaches its highest value at K=4, representing the best intra-cluster similarity and inter-cluster difference at this point.

Combining these two indicators, K=4 is a more reasonable and interpretable cluster number, which can be used to further analyze the different patterns between 'Center-Based care vs. Family Childcare dominance' and the Labor Force Participation Rate (LFPR), and to pinpoint counties and cities that may lack flexible care support.

```python
# output cluster profile

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

k_final = 4
df_q2['cluster_kmeans'] = labels_by_k[k_final]

print('K-means cluster sizes (k=4):')
display(df_q2['cluster_kmeans'].value_counts().sort_index())  # cluster size

# take the mean for each feature
profile_raw = (df_q2.groupby('cluster_kmeans')[feature_cols].mean().round(2))
print('Cluster profile (raw means):')
display(profile_raw)

# standardize: convert values to z-score (more comparable)
df_std = df_q2.copy()
for col in feature_cols:
    df_std[col] = (df_std[col] - df_std[col].mean()) / df_std[col].std(ddof=0)

profile_std = (df_std.groupby('cluster_kmeans')[feature_cols].mean().round(2))  # take the mean after standardization
print('Cluster profile (standardized means):')
display(profile_std)

# plot heatmap
plt.figure(figsize=(10, 4))
sns.heatmap(profile_std, annot=True, fmt='.2f', center=0, cmap='coolwarm')
plt.title('K-means Cluster Profiles (standardized means, k=4)')
plt.xlabel('Features')
plt.ylabel('Cluster')
plt.show()
```

The clustering reveals meaningful heterogeneity across counties, where certain childcare cost and provider dominance patterns are associated with lower female labor force participation, indicating potential gaps in childcare infrastructure that may constrain flexible work participation.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# scatterplot: provider advantage vs female LFP, colored by cluster
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df_q2, x='provider_advantage', y='flfpr_20to64', hue='cluster_kmeans', alpha=0.7)
plt.axvline(0, linestyle='--', linewidth=1)  # the left and right sides of 0 represent which provider is more dominant
plt.title('Provider Advantage vs Female LFPR')
plt.xlabel('Provider advantage (Family Childcare - Center-Based care)')
plt.ylabel('Female LFPR (20–64)')
plt.legend(title='Cluster', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

The clustering reveals heterogeneous patterns in how provider dominance relates to female labor force participation. Some counties show persistently low participation despite more affordable family childcare, suggesting potential gaps in childcare infrastructure and schedule flexibility.
